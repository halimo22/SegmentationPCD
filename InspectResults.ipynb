# Simple code to transform Numpy array (.npy) into 3D object for inspection (.obj)

```python
import numpy as np

def class_to_color(class_labels):
    """Map class labels to specific colors."""
    color_map = {
        0: [0, 0, 255],      # Ceiling - Blue
        1: [0, 255, 0],      # Floor - Green
        2: [255, 0, 0],      # Wall - Red
        3: [255, 255, 0],    # Beam - Yellow
        4: [255, 165, 0],    # Column - Orange
        5: [0, 255, 255],    # Window - Cyan
        6: [128, 128, 128],  # Door - Gray
        7: [255, 20, 147],   # Table - Pink
        8: [128, 0, 128],    # Chair - Purple
        9: [165, 42, 42],    # Sofa - Brown
        10: [0, 128, 128],   # Bookcase - Teal
        11: [255, 105, 180], # Board - Light Pink
        12: [192, 192, 192]  # Clutter - Silver
    }
    default_color = [255, 255, 255]  # White
    colors = np.array([color_map.get(cls, default_color) for cls in class_labels])
    return colors / 255.0  # Normalize to [0, 1]


def npy_to_obj(points_file, labels_file, obj_file):
    """Convert two .npy files (points and class labels) to an OBJ file."""
    # Load points (Assume Nx3: x, y, z)
    points = np.load(points_file).astype(float)

    # Load the dictionary from the labels file
    labels_dict = np.load(labels_file, allow_pickle=True).item()

    # Extract the 'predict_labels' key
    class_labels = labels_dict['predict_labels']

    if len(points) != len(class_labels):
        raise ValueError("Mismatch between number of points and class labels.")

    # Map class labels to colors
    colors = class_to_color(class_labels)

    # Write to an OBJ file
    with open(obj_file, 'w') as f:
        for point, color in zip(points, colors):
            f.write(f"v {point[0]} {point[1]} {point[2]} {color[0]} {color[1]} {color[2]}\n")
    print(f"OBJ file saved to {obj_file}")


# Example usage
points_file = "/home/seif/Downloads/points.npy"  # Path to your .npy file containing points
labels_file = "/home/seif/Downloads/predictions.npy"  # Path to your .npy file containing labels
obj_file = "output2.obj"     # Path to save the output OBJ file
npy_to_obj(points_file, labels_file, obj_file)
```
