```python
import os
import cv2
import pandas as pd
from ultralytics import YOLO
from IPython.display import clear_output

model = YOLO("runs/detect/train9/weights/best.pt")

train_images_path = "T.E.A---TestesElementosArquitetonicos-4/train/images"
val_images_path = "T.E.A---TestesElementosArquitetonicos-4/valid/images"

cropped_train_dir = "cropped/train"
cropped_val_dir = "cropped/val"
os.makedirs(cropped_train_dir, exist_ok=True)
os.makedirs(cropped_val_dir, exist_ok=True)

CONF_THRESHOLD = 0.5  

cropped_data = []

def process_images(images_path, output_dir, dataset_type):
    image_files = [f for f in os.listdir(images_path) if f.endswith(('.jpg', '.png', '.jpeg'))]

    for index, image_file in enumerate(image_files, start=1):
        image_path = os.path.join(images_path, image_file)
        results = model(image_path, conf=CONF_THRESHOLD)
        img = cv2.imread(image_path)

        num_objects = 0
        
        for i, result in enumerate(results):
            boxes = result.boxes
            for j, box in enumerate(boxes):
                x1, y1, x2, y2 = map(int, box.xyxy[0])
                class_id = int(box.cls[0].item())
                class_name = model.names[class_id]
                
                cropped_img = img[y1:y2, x1:x2]
                
                cropped_filename = f"{dataset_type}_{image_file.split('.')[0]}_{j}.jpg"
                cropped_filepath = os.path.join(output_dir, cropped_filename)
                cv2.imwrite(cropped_filepath, cropped_img)
                
                cropped_data.append([cropped_filename, class_name, image_file, x1, y1, x2, y2])
                num_objects += 1
        
        print(f"Processed {image_file}: Detected {num_objects} objects.")
        if num_objects > 0:
            print(f"    Cropped images saved in {output_dir}.")
        
        if index % 50 == 0:
            clear_output(wait=True)

print("Processing training images...")
process_images(train_images_path, cropped_train_dir, "train")

print("Processing validation images...")
process_images(val_images_path, cropped_val_dir, "val")

csv_path = "cropped_data.csv"
df = pd.DataFrame(cropped_data, columns=["Cropped Image", "Class", "Original Image", "x1", "y1", "x2", "y2"])
df.to_csv(csv_path, index=False)

clear_output(wait=True)
print("\nProcessing complete!")
print(f"Cropped images saved in {cropped_train_dir} and {cropped_val_dir}.")
print(f"CSV file generated: {csv_path}")
```

```python
import torch
import torch.nn as nn
from torchvision.models import efficientnet_b7

class EfficientNetB7Classifier(nn.Module):
    def __init__(self, num_classes=8):
        super(EfficientNetB7Classifier, self).__init__()
        self.model = efficientnet_b7(pretrained=True)
        num_features = self.model.classifier[1].in_features
        self.model.classifier = nn.Sequential(
            nn.Linear(num_features, 128),
            nn.ReLU(),
            nn.Dropout(0.4),
            nn.Linear(128, num_classes)
        )
    
    def forward(self, x):
        return self.model(x)

# Move model to GPU if available
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = EfficientNetB7Classifier().to(device)
```

```python
import os
import pandas as pd
import shutil
from sklearn.model_selection import train_test_split
import numpy as np

# Load the CSV file
df = pd.read_csv("cropped_data_limited.csv")

# Split dataset into train (90%) and validation (10%)
train_df, val_df = train_test_split(df, test_size=0.1, stratify=df["Class"], random_state=42)

# Create directories
os.makedirs("dataset/train", exist_ok=True)
os.makedirs("dataset/val", exist_ok=True)

# Function to copy images to respective directories
def copy_images(dataframe, dataset_type):
    for _, row in dataframe.iterrows():
        class_dir = os.path.join(f"dataset/{dataset_type}", row["Class"])
        os.makedirs(class_dir, exist_ok=True)
        src_path = os.path.join("cropped", row["Cropped Image"])
        dst_path = os.path.join(class_dir, row["Cropped Image"])
        shutil.copy(src_path, dst_path)

copy_images(train_df, "train")
copy_images(val_df, "val")
```

```python
from torchvision.datasets import ImageFolder
from torch.utils.data import DataLoader
import torchvision.transforms as transforms

transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])
# Load datasets
train_dataset = ImageFolder("dataset/train", transform=transform)
val_dataset = ImageFolder("dataset/val", transform=transform)

train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=32, shuffle=False)
```

```python
import os
import pandas as pd
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision.models import efficientnet_b7
from torch.optim.lr_scheduler import ReduceLROnPlateau
import numpy as np


# Define loss function, optimizer, and learning rate scheduler
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)
scheduler = ReduceLROnPlateau(optimizer, mode='min', factor=0.1, patience=3, verbose=True)

# Training parameters
num_epochs = 20
early_stopping_patience = 5
best_val_loss = np.inf
patience_counter = 0

# Training loop
for epoch in range(num_epochs):
    model.train()
    running_loss = 0.0
    for images, labels in train_loader:
        images, labels = images.to(device), labels.to(device)
        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        running_loss += loss.item()
    
    # Validation phase
    model.eval()
    val_loss = 0.0
    correct = 0
    total = 0
    with torch.no_grad():
        for images, labels in val_loader:
            images, labels = images.to(device), labels.to(device)
            outputs = model(images)
            loss = criterion(outputs, labels)
            val_loss += loss.item()
            _, predicted = torch.max(outputs, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()
    
    val_acc = correct / total
    avg_train_loss = running_loss / len(train_loader)
    avg_val_loss = val_loss / len(val_loader)
    
    print(f"Epoch {epoch+1}/{num_epochs}, Train Loss: {avg_train_loss:.4f}, Val Loss: {avg_val_loss:.4f}, Val Acc: {val_acc:.4f}")
    
    scheduler.step(avg_val_loss)
    
    # Checkpoint saving
    if avg_val_loss < best_val_loss:
        best_val_loss = avg_val_loss
        patience_counter = 0
        torch.save(model.state_dict(), "best_model.pth")
        print("Model saved!")
    else:
        patience_counter += 1
    
    # Early stopping
    if patience_counter >= early_stopping_patience:
        print("Early stopping triggered.")
        break

print("Training complete.")
```

```python
import torch
import torchvision.transforms as transforms
from torchvision.models import efficientnet_b7
from PIL import Image
import cv2
import numpy as np
import os
import matplotlib.pyplot as plt
import pandas as pd



image_path='/scratch/project_2012411/Scan6Images/Extracted/NewOffset/ObjectforScan06NewOffset/cropped_image_19.jpg'
# Load trained weights
model = EfficientNetB7Classifier()  # Ensure model is defined
model.load_state_dict(torch.load("best_model.pth", map_location=torch.device("cpu")))
model.eval()

# Define image transformation for classification
transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

# Apply Rectilinear Projection Correction
def rectilinear_projection(image_path):
    img = cv2.imread(image_path)
    h, w = img.shape[:2]
    
    # Define focal length based on image width (adjustable parameter)
    focal_length = w
    K = np.array([[focal_length, 0, w/2], [0, focal_length, h/2], [0, 0, 1]])
    
    # Define distortion coefficients (assuming moderate fisheye distortion)
    dist_coeffs = np.array([-0.2, 0.05, 0, 0])
    
    # Undistort image using perspective transformation
    rectified_img = cv2.undistort(img, K, dist_coeffs)
    return rectified_img

# Correct the image before classification
corrected_image = rectilinear_projection(image_path)
cv2.imwrite("corrected_image.jpg", corrected_image)

# Load and preprocess corrected image for classification
image = Image.open("corrected_image.jpg").convert("RGB")
original_image = Image.open("corrected_image.jpg")
image = transform(image).unsqueeze(0)

# Perform inference
with torch.no_grad():
    output = model(image)
    _, predicted_class = torch.max(output, 1)
    predicted_label = class_names[predicted_class.item()]

# Display the corrected image
plt.figure(figsize=(6,6))
plt.imshow(original_image)
plt.axis("off")
plt.title(f"Predicted class: {predicted_label}")
plt.show()

# Print classification result
print(f"Predicted class: {predicted_label}")
```
